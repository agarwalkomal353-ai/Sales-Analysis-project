### Target vs Actual Sales (Category-wise)¶
Objective :
To compare the actual sales with the sales target for each product category during the period where target data is available (April 2018 to March 2019).

```python
# Filter data where target is available

target_data = final_data[final_data['target'].notna()]

print(target_data.shape)
```

```python
# Actual Sales
actual_sales = (
    target_data.groupby('category', as_index=False)['total_sales']
    .sum()
)

# Target Sales (remove duplicate month-category targets)
target_sales = (
    target_data[['month_of_order_date', 'category', 'target']]
    .drop_duplicates()
    .groupby('category', as_index=False)['target']
    .sum()
)

# Merge both
target_vs_sales = actual_sales.merge(target_sales, on='category')

display(target_vs_sales)
```

### **Handling Duplicate Target Values After Merge**

After merging the datasets, the **target** values were repeated for every order belonging to the same **month** and **category**. This happened because one target record matched multiple order records (a one-to-many relationship).

If the target values were summed directly, the same monthly target would be counted multiple times, resulting in incorrect and inflated target values.

To avoid this issue, duplicate combinations of **month_of_order_date**, **category**, and **target** were removed using `drop_duplicates()`. The target values were then aggregated, ensuring that each monthly target was counted only once before comparing it with the actual sales.


```python
# Calculate Target Achievement (%)

target_vs_sales['achievement_%'] = (
    target_vs_sales['total_sales'] / target_vs_sales['target']
) * 100

target_vs_sales['achievement_%'] = target_vs_sales['achievement_%'].round(2)

display(target_vs_sales)
```

```python
# Target Achievement by Category (%)

plt.figure(figsize=(7,5))

sns.barplot(
    data=target_vs_sales,
    x='category',
    y='achievement_%'
)

plt.title("Target Achievement by Category (%)")
plt.xlabel("Category")
plt.ylabel("Achievement (%)")

plt.show()
```

### **Observation**

* **Electronics** achieved the highest target achievement (**30.07%**).
* **Furniture** ranked second with a target achievement of **25.85%**.
* **Clothing** recorded the lowest target achievement (**20.58%**).
* None of the product categories achieved **50%** of their assigned sales target during the target period.
